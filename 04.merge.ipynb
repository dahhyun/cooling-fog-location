```python
import pandas as pd
from google.colab import files

df1 = pd.read_csv("/content/성환읍_후보지(MCLP).csv",  encoding="cp949")
df2 = pd.read_csv("/content/성환읍_후보지(P-median).csv",  encoding="cp949")

# 공통 컬럼 기준 교집합 merge
merged = pd.merge(df1, df2, on="이름", how="inner")

# 중복 제거
merged_unique = merged.drop_duplicates()

# 저장
merged_unique.to_csv("/content/merged_file(성환읍_수정).csv", index=False, encoding="utf-8-sig")
```
