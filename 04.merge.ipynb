```python
import pandas as pd
df = pd.read_csv("입지후보지.csv")

df1 = pd.read_csv("성환읍_후보지(MCLP).csv")
df2 = pd.read_csv("성환읍_후보지(P-median).csv")

# 공통 컬럼 기준 교집합 merge
merged = pd.merge(df1, df2, on="이름", how="inner")

# 중복 제거
merged_unique = merged.drop_duplicates()
new_df1 = merged_unique.iloc[:,:3]

new_df1 = new_df1.merge(df[["이름", "주소"]], on="이름", how="left").drop_duplicates()
new_df1['행정동'] = '성환읍'
new_df1
```

```python
df1 = pd.read_csv("청룡동_후보지(MCLP).csv")
df2 = pd.read_csv("청룡동_후보지(P-median).csv")

# 공통 컬럼 기준 교집합 merge
merged = pd.merge(df1, df2, on="이름", how="inner")

# 중복 제거
merged_unique = merged.drop_duplicates()
new_df2 = merged_unique.iloc[:,:3]

new_df2 = new_df2.merge(df[["이름", "주소"]], on="이름", how="left").drop_duplicates()
new_df2['행정동'] = '청룡동'
new_df2.drop(2, axis=0, inplace=True)
new_df2.reset_index(drop=True)
```

```python
df1 = pd.read_csv("신안동_후보지(MCLP).csv")
df2 = pd.read_csv("신안동_후보지(P-median).csv")

# 공통 컬럼 기준 교집합 merge
merged = pd.merge(df1, df2, on="이름", how="inner")

# 중복 제거
merged_unique = merged.drop_duplicates()
new_df3 = merged_unique.iloc[:,:3]

new_df3 = new_df3.merge(df[["이름", "주소"]], on="이름", how="left").drop_duplicates()
new_df3['행정동'] = '신안동'
new_df3.drop(1, axis=0, inplace=True)
new_df3.reset_index(drop=True)
```

```python
result = pd.concat([new_df1, new_df2, new_df3]).reset_index(drop=True)
result.columns = ['시설명', '위도', '경도', '주소', '행정동']
```
